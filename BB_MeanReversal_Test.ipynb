```python
import importlib
from ib_async import IB
import Backtesting.DataLoader.ibkr_historical_loader as ibdl

importlib.reload(ibdl)

TICKERS = [
    "LRCX",
    "MU",
    "AMAT",
    "INTC",
    "PLTR",
    "XOM",
    "GOOG",
    "SPY",
    "META",
    "NVDA",
]

START = "2016-10-03"
END   = "2026-10-02"

HOST = "127.0.0.1"
PORT = 4002
CLIENT_ID = 56

ib = IB()

await ib.connectAsync(
    HOST,
    PORT,
    clientId=CLIENT_ID,
    timeout=15,
)

print("Connected:", ib.isConnected())

results = {}

try:
    for ticker in TICKERS:
        print("\n" + "=" * 100)
        print(f"DOWNLOADING {ticker} | 4H | {START} -> {END}")
        print("=" * 100)

        df = await ibdl.download_data_ibkr_prepare_csv_cache(
            ticker=ticker,
            start=START,
            end=END,
            interval="4h",
            use_rth=True,
            ib=ib,
            disconnect_when_done=False,
            pacing_sleep_seconds=0.5,
        )

        results[ticker] = df

        print(
            f"{ticker}: rows={len(df)} | "
            f"{df.index.min() if len(df) else 'NO DATA'} -> "
            f"{df.index.max() if len(df) else 'NO DATA'}"
        )

finally:
    if ib.isConnected():
        ib.disconnect()

print("\nALL DONE")
```
