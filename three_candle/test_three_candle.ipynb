# Rudra Three-Candle — NQ=F 15-minute signals
Run cells in order. Begin with **one day**, then extend the dates. Charts and printed candles are for review; no trade execution is simulated.

```python
%pip install -q yfinance pandas matplotlib
import sys
from pathlib import Path
repo = Path.cwd()
module_dir = repo / 'three_candle' if (repo / 'three_candle').exists() else repo
sys.path.insert(0, str(module_dir))
from three_candle import download_nq, find_signals, print_signal, plot_signal
import matplotlib.pyplot as plt
```

```python
# EDIT THESE TWO DATES: start with one day, then use a whole month.
START = '2026-10-29'
END = '2026-10-29'
REQUIRE_BREAKOUT = False
data = download_nq(START, END)
print(f'{len(data)} candles downloaded, in UK time')
display(data.head())
```

```python
signals = list(find_signals(data, require_breakout=REQUIRE_BREAKOUT))
print(f'{len(signals)} setups found')
for number, signal in enumerate(signals, start=1):
    print(number, signal.time, signal.direction, 'C2 high:', signal.candle2_high)
```

```python
# Change SIGNAL_NUMBER to inspect signals individually (1-based).
SIGNAL_NUMBER = 1
if signals:
    signal = signals[SIGNAL_NUMBER - 1]
    print_signal(signal, data)
    fig = plot_signal(signal, data, before=12, after=8)
    display(fig)
    plt.close(fig)
else:
    print('No signals on selected dates; try a longer date range.')
```

```python
# Optional: print every setup's three candles without opening charts.
for signal in signals:
    print_signal(signal, data)
```
